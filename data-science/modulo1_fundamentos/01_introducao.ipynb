## Mediana

Ordene os valores. A mediana é o valor central; se a quantidade for par, tire a média dos dois centrais. Ela é menos sensível a extremos do que a média.

**Ponte com SQL:** ORDER BY ajuda a ordenar, mas não calcula a mediana sozinho. Funções de percentil variam entre bancos.



````python
# Exemplo: Mediana
valores = sorted([10, 90, 20, 30])
meio = len(valores) // 2
print((valores[meio - 1] + valores[meio]) / 2)
````

**Resultado esperado:**

## Moda

Moda é o valor mais frequente. Pode haver empate entre várias modas. Quando todos aparecem uma vez, adotaremos a convenção de dizer que não há moda.

**Ponte com SQL:** GROUP BY valor com COUNT(*) resume frequências. Não escolha uma única linha se houver empate na maior frequência.



````python
# Exemplo: Moda
valores = [2, 2, 3, 4]
frequencias = {}
for valor in valores:
    frequencias[valor] = frequencias.get(valor, 0) + 1
maior = max(frequencias.values())
modas = sorted([v for v in frequencias if frequencias[v] == maior])
print(modas if maior > 1 else [])
````

**Resultado esperado:**
```text
[2]
```

## Variância

A variância mede a dispersão por meio dos desvios quadráticos em relação à média. Para descrever toda a população, dividimos por n. Na estimativa amostral usual, dividimos por n − 1.

**Ponte com SQL:** VAR_POP e VAR_SAMP, quando disponíveis no banco, usam essas duas convenções. Compare sempre a mesma definição.

Os exemplos pressupõem dados numéricos não vazios. Variância amostral precisa de pelo menos dois valores.

````python
# Exemplo: Variância
valores = [2, 4, 6]
media = sum(valores) / len(valores)
variancia = sum((v - media) ** 2 for v in valores) / len(valores)
print(round(variancia, 2))
````

**Resultado esperado:**
```text
2.67
```

## Desvio padrão

É a raiz quadrada da variância e volta à unidade original: se os valores são reais, o desvio padrão também é expresso em reais.

**Ponte com SQL:** STDDEV_POP e STDDEV_SAMP dependem do banco. Assim como na variância, verifique qual convenção está usando.



````python
# Exemplo: Desvio padrão
valores = [2, 4]
media = sum(valores) / len(valores)
variancia = sum((v - media) ** 2 for v in valores) / len(valores)
print(variancia ** 0.5)
````

Não generalize os índices deste exemplo para listas de outros tamanhos.

## Probabilidade

Em resultados igualmente prováveis, a probabilidade é a razão entre casos favoráveis e possíveis. Frequências observadas estimam probabilidades, mas não garantem o próximo resultado.

**Ponte com SQL:** SUM(CASE WHEN evento THEN 1 ELSE 0 END) dividido pelo total calcula uma frequência relativa; cuide da divisão inteira e do denominador.



````python
# Exemplo: Probabilidade
faces = [1, 2, 3, 4, 5, 6]
pares = [v for v in faces if v % 2 == 0]
print(len(pares) / len(faces))
````

**Resultado esperado:**

## Correlação

A correlação de Pearson mede associação linear entre duas variáveis e varia de −1 a 1. Perto de zero não significa ausência de toda relação. Variáveis constantes tornam o coeficiente indefinido.

**Ponte com SQL:** CORR(x, y), onde disponível, trabalha com pares válidos. Mantenha o alinhamento das observações ao filtrar ou fazer joins.

A fórmula pressupõe pares numéricos, listas do mesmo tamanho e variabilidade nos dois eixos.

````python
# Exemplo: Correlação
x = [1, 2, 3]
y = [2, 4, 6]
mx, my = sum(x) / len(x), sum(y) / len(y)
numerador = sum((a - mx) * (b - my) for a, b in zip(x, y))
denominador = (sum((a - mx) ** 2 for a in x) * sum((b - my) ** 2 for b in y)) ** 0.5
print(round(numerador / denominador, 2))
````

**Resultado esperado:**
```text
1.0
```

## Laços

for percorre elementos. while repete enquanto uma condição for verdadeira e precisa poder terminar. Prefira exemplos pequenos para observar cada passo.

**Ponte com SQL:** SQL descreve operações sobre conjuntos. Um laço é uma ferramenta diferente, não uma tradução obrigatória de cada consulta.



````python
# Exemplo: Laços
total = 0
for valor in [10, 20, 30]:
    total += valor
print(total)
contador = 0
while contador < 2:
    contador += 1
print(contador)
````

**Resultado esperado:**
```text
60
2
```

## Funções
